```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import warnings 
warnings.filterwarnings('ignore')
```

```python
# IPL 2022 Capstone Project
# The Indian Premier League (IPL) is a professional T20 cricket league in India, featuring franchises representing cities. This project explores IPL 2022 match-level data to derive meaningful insights and understand match outcomes, player performances, and team dynamics.

# These are some of the important columns that we'll focus on for meaningful insights in this project.

# column names: Variable Type

# date : string
# venue : string
# stage : string
# team1 : string
# team2 : string
# toss_winner : string
# toss_decision : string
# first_ings_score : integer
# second_ings_score : integer
# match_winner : string
# won_by : string
# margin : integer
# player_of_the_match : string
# top_scorer : string
# highscore : integer
# best_bowling : string
# best_bowling_fgure : string
# gure : string
```

```python
df = pd.read_csv(r'IPL.csv')
df
```

```python
#checking if is there any null value exist
df.isna()
df.isna().sum()
```

```python
#which team won the most matches
most_matchwined = df['match_winner'].value_counts()
sns.barplot(y = most_matchwined.index, x= most_matchwined.values,palette='viridis')
plt.title('MOST MATCHED WINS TEAM')
```

```python
#TOSS DISICION TRENDS
sns.countplot(x= df['toss_decision'],palette='rainbow')
plt.title('TOSS DESICION TRENDS')
```

```python
#toss winners vs match winners if team win the toss so what it won the match 
count = df[df['toss_winner'] == df['match_winner']]['match_id'].count()
perct = (count * 100 ) /df.shape[0]
perct.round(2)
```

```python
# how do teams win by wickets or by runs
abc = df['won_by'].value_counts()
abc
#same as upper code but now visually
sns.countplot(x= df['won_by'])
```

```python
#most player_of_the_match award winner
#sns.countplot(x= )
count = df['player_of_the_match'].value_counts().head(10)
count
sns.barplot(x=count.values,y=count.index)
```

```python
#top 2 top scorrer
ts= df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
ts
ts.plot(kind='barh')
```

```python
#10 best bowling figures
df['highest_wickets'] = df['best_bowling_figure'].apply(lambda x: x.split('--')[0] )
df['highest_wickets']= df['highest_wickets'].astype(int) 
top_bowlers =df.groupby('best_bowling')['highest_wickets'].sum().sort_values(ascending=False).head(10)
top_bowlers
top_bowlers.plot(kind='barh')
```

```python
#most match played by vanue 
v_count = df['venue'].value_counts()
v_count
v_count.plot(kind='barh')
```

```python
#who wans the higest margin by runs
df[df['won_by'] == 'Runs'].sort_values(by='margin',ascending=False).head(1)[['match_winner','margin']]
```

```python
#which player have hogest individual score
df[df['highscore'] == df['highscore'].max()][['top_scorer','highscore']]
```

```python
#bowler who had higest wickets
df[df['highest_wickets'] == df['highest_wickets'].max()][['best_bowling','best_bowling_figure']]
```
